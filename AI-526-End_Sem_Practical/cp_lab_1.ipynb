```python
import numpy as np
from numba import cuda

# Matrix dimensions
N = 1024
scalar = 10

# Input matrix
A = np.random.randint(0, 100, (N, N)).astype(np.float32)

# Output matrix
C = np.zeros_like(A)

# CUDA Kernel
@cuda.jit
def scalar_add_kernel(A, scalar, C):

    row, col = cuda.grid(2)

    if row < A.shape[0] and col < A.shape[1]:
        C[row, col] = A[row, col] + scalar

# Threads and blocks
threads_per_block = (16, 16)

blocks_x = (N + threads_per_block[0] - 1) // threads_per_block[0]
blocks_y = (N + threads_per_block[1] - 1) // threads_per_block[1]

blocks_per_grid = (blocks_x, blocks_y)

# Launch Kernel
scalar_add_kernel[blocks_per_grid, threads_per_block](A, scalar, C)

cuda.synchronize()

print("Input Matrix Sample:")
print(A[:5,:5])

print("\nOutput Matrix Sample:")
print(C[:5,:5])
```
