## **1) Residual Block and Mini-ResNet on CIFAR-10**

```python
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.datasets import cifar10
from tensorflow.keras.utils import to_categorical
import matplotlib.pyplot as plt

# Load CIFAR-10 Dataset
(x_train, y_train), (x_test, y_test) = cifar10.load_data()

# Normalize images
x_train = x_train / 255.0
x_test = x_test / 255.0

# One-hot encoding
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

# ---------------------------------------------------
# Build Residual Block Manually
# ---------------------------------------------------
def residual_block(input_tensor, filters):

    # Save input tensor (Skip Connection)
    shortcut = input_tensor

    # First Convolution Layer
    x = layers.Conv2D(
        filters=filters,
        kernel_size=(3,3),
        padding='same'
    )(input_tensor)

    x = layers.BatchNormalization()(x)
    x = layers.ReLU()(x)

    # Second Convolution Layer
    x = layers.Conv2D(
        filters=filters,
        kernel_size=(3,3),
        padding='same'
    )(x)

    x = layers.BatchNormalization()(x)

    # Add Skip Connection
    x = layers.Add()([x, shortcut])

    # Activation
    x = layers.ReLU()(x)

    return x

# ---------------------------------------------------
# Build Mini ResNet Model
# ---------------------------------------------------
inputs = layers.Input(shape=(32,32,3))

# Initial Convolution Layer
x = layers.Conv2D(
    32,
    (3,3),
    padding='same',
    activation='relu'
)(inputs)

# Residual Block 1
x = residual_block(x, 32)

# Residual Block 2
x = residual_block(x, 32)

# Max Pooling
x = layers.MaxPooling2D((2,2))(x)

# Additional Convolution Layer
x = layers.Conv2D(
    64,
    (3,3),
    activation='relu',
    padding='same'
)(x)

# Max Pooling
x = layers.MaxPooling2D((2,2))(x)

# Flatten Layer
x = layers.Flatten()(x)

# Fully Connected Layer
x = layers.Dense(128, activation='relu')(x)

# Output Layer
outputs = layers.Dense(10, activation='softmax')(x)

# Create Model
model = models.Model(inputs, outputs)

# ---------------------------------------------------
# Display Model Summary
# ---------------------------------------------------
model.summary()

# ---------------------------------------------------
# Compile Model
# ---------------------------------------------------
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# ---------------------------------------------------
# Train Mini ResNet Model
# ---------------------------------------------------
history = model.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=64,
    validation_split=0.2
)

# ---------------------------------------------------
# Evaluate Model
# ---------------------------------------------------
test_loss, test_accuracy = model.evaluate(x_test, y_test)

print("\nTest Accuracy :", test_accuracy)

# ---------------------------------------------------
# Plot Accuracy Graph
# ---------------------------------------------------
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')

plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.title('Mini ResNet on CIFAR-10')

plt.legend()
plt.show()
```

## **2) Adjustable Stride and Padding in CNN**

```python
import tensorflow as tf
import numpy as np

# Create random input image
# Shape = (Batch Size, Height, Width, Channels)
input_image = np.random.rand(1, 28, 28, 1).astype(np.float32)

# Function to test convolution layer
def convolution_demo(stride_value, padding_type):

    # Create Conv Layer
    conv = tf.keras.layers.Conv2D(
        filters=1,
        kernel_size=(3,3),
        strides=stride_value,
        padding=padding_type
    )

    # Apply convolution
    output = conv(input_image)

    # Print results
    print("\n--------------------------------")
    print("Stride :", stride_value)
    print("Padding:", padding_type)
    print("Output Shape:", output.shape)
    print("--------------------------------")

# Different combinations
convolution_demo(1, 'valid')
convolution_demo(2, 'valid')
convolution_demo(1, 'same')
convolution_demo(2, 'same')
```

## **3) CNN with Regularization and Dropout on Fashion-MNIST**

```python
import tensorflow as tf
from tensorflow.keras import layers, models, regularizers
from tensorflow.keras.datasets import fashion_mnist
from tensorflow.keras.utils import to_categorical

# ---------------------------------------------------
# Load Fashion MNIST Dataset
# ---------------------------------------------------
(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()

# Normalize
x_train = x_train / 255.0
x_test = x_test / 255.0

# Reshape for CNN
x_train = x_train.reshape(-1, 28, 28, 1)
x_test = x_test.reshape(-1, 28, 28, 1)

# One-hot encoding
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

# ---------------------------------------------------
# Function to create CNN model
# ---------------------------------------------------
def create_model(reg=None, dropout=False):

    model = models.Sequential()

    # 1st Conv Layer
    model.add(layers.Conv2D(
        32,
        (3,3),
        activation='relu',
        padding='same',
        kernel_regularizer=reg,
        input_shape=(28,28,1)
    ))

    # 2nd Conv Layer
    model.add(layers.Conv2D(
        32,
        (3,3),
        activation='relu',
        padding='same',
        kernel_regularizer=reg
    ))

    model.add(layers.MaxPooling2D((2,2)))

    # 3rd Conv Layer
    model.add(layers.Conv2D(
        64,
        (3,3),
        activation='relu',
        padding='same',
        kernel_regularizer=reg
    ))

    # 4th Conv Layer
    model.add(layers.Conv2D(
        64,
        (3,3),
        activation='relu',
        padding='same',
        kernel_regularizer=reg
    ))

    model.add(layers.MaxPooling2D((2,2)))

    # Dropout Layer
    if dropout:
        model.add(layers.Dropout(0.5))

    model.add(layers.Flatten())

    # Dense Layer
    model.add(layers.Dense(
        128,
        activation='relu',
        kernel_regularizer=reg
    ))

    # Another Dropout
    if dropout:
        model.add(layers.Dropout(0.5))

    # Output Layer
    model.add(layers.Dense(10, activation='softmax'))

    # Compile Model
    model.compile(
        optimizer='adam',
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )

    return model

# ---------------------------------------------------
# Create Different Models
# ---------------------------------------------------

# a) Base Model
base_model = create_model()

# b) L1 Regularization Model
l1_model = create_model(
    reg=regularizers.l1(0.001)
)

# c) L2 Regularization Model
l2_model = create_model(
    reg=regularizers.l2(0.001)
)

# d) Dropout Model
dropout_model = create_model(
    dropout=True
)

# e) L2 + Dropout Model
combined_model = create_model(
    reg=regularizers.l2(0.001),
    dropout=True
)

# ---------------------------------------------------
# Store models
# ---------------------------------------------------
models_dict = {
    "Base Model": base_model,
    "L1 Model": l1_model,
    "L2 Model": l2_model,
    "Dropout Model": dropout_model,
    "L2 + Dropout Model": combined_model
}

# ---------------------------------------------------
# Train and Evaluate Models
# ---------------------------------------------------
results = {}

for name, model in models_dict.items():

    print("\n===================================")
    print("Training:", name)
    print("===================================")

    history = model.fit(
        x_train,
        y_train,
        epochs=5,
        batch_size=64,
        validation_split=0.2,
        verbose=1
    )

    # Training Accuracy
    train_acc = history.history['accuracy'][-1]

    # Test Accuracy
    test_loss, test_acc = model.evaluate(x_test, y_test)

    # Store results
    results[name] = (train_acc, test_acc)

# ---------------------------------------------------
# Final Results
# ---------------------------------------------------
print("\n\n========== FINAL ACCURACY ==========")

for name, (train_acc, test_acc) in results.items():

    print("\n", name)
    print("Training Accuracy :", round(train_acc, 4))
    print("Test Accuracy     :", round(test_acc, 4))
```
